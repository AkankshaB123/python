# Use .split(), .join() to convert the list_of_names to the output

```python
list_of_names = ['Peter Parker', 'Clark Kent', 'Wade Wilson', 'Bruce Wayne']

output = ['PP', 'CK', 'WW', 'BW']
```

```python
output = []
for name in list_of_names:
    first_letter_list = []
    for word in name.split(" "):
         first_letter_list.append(word[0])
    output.append("".join(first_letter_list))
```

```python
output
```

## Step-by-Step Explanation:
* Iterated through each and every name
* I initialised an empty list
* By splitting the name into multiple words on the basis of space (" ")
* I collected the first letter of each word and appended to first letter list
* Joined them using no character "" and appended it to output
