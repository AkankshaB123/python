```python
def consecutive_numbers(logs: pd.DataFrame) -> pd.DataFrame:
  logs = logs.sort_values(by=['id'], ascending=True)
  logs['num_shift'] = logs['num'].shift(1)
  logs['num_shift_2'] = logs['num'].shift(2)

  # Filter for rows where current, previous, and two-previous numbers are the same
  consecutive_filter = ((logs['num'] == logs['num_shift']) &
                        (logs['num_shift'] == logs['num_shift_2']))

  # Get the numbers that meet the criteria and convert to DataFrame of unique values
  consecutive_nums_series = logs.loc[consecutive_filter, 'num']
  result_df = pd.DataFrame({'ConsecutiveNums': consecutive_nums_series.unique()})

  return result_df

display(consecutive_numbers(df))
print(type(df))
```
