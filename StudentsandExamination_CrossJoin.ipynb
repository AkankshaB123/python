```python
import pandas as pd

students = pd.DataFrame({
    "student_id": [1, 2, 13, 6],
    "student_name": ["Alice", "Bob", "John", "Alex"]
})

subjects = pd.DataFrame({
    "subject_name": ["Math", "Physics", "Programming"]
})

examinations = pd.DataFrame({
    "student_id": [1, 1, 1, 2, 1, 1, 13, 13, 13, 2, 1],
    "subject_name": [
        "Math", "Physics", "Programming", "Programming",
        "Physics", "Math", "Math", "Programming",
        "Physics", "Math", "Math"
    ]
})
```

```python
examinations['attended_exams'] = 1
df = pd.merge(students, subjects, how='cross')
df_final = pd.merge(df, examinations, on=['student_id', 'subject_name'], how='left')
df_final
```

```python
df_final.groupby(['student_id','student_name','subject_name'])['attended_exams'].count().reset_index()
```
