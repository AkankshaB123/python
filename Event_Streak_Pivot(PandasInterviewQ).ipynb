Similar interview question
1. Find average session duration for each user.
2. Calculate machine running time from Start/Stop logs. (Real-company scenario eg. Databricks usage)
3. Find order processing time (Order Created -> Order Delivered)
4. Compute employee clock-in/clock-out hours

Approach 1: 1) case when status = 'start' + group by project and operation to get timestamp
Approach 2: 1) Row_number() partition by project and operation, order by timestamp asc to get timestamp start
Approch 3: 1) Self-Join (Esp. used in Python case to separate start and end.

```python
records = pd.DataFrame([
    {"project_id": 1, "operation_id": 101, "status": "Start", "timestamp": 26276},
    {"project_id": 1, "operation_id": 101, "status": "End", "timestamp": 48494},
    {"project_id": 2, "operation_id": 102, "status": "Start", "timestamp": 50000},
    {"project_id": 2, "operation_id": 102, "status": "End", "timestamp": 55000},
])
```

```python
import pandas as pd
```

```python
def fetch_avg_timestamp_per_grp(trans_tbl: pd.DataFrame) -> pd.DataFrame:
  start_grp = trans_tbl[trans_tbl['status'] == 'Start']
  end_grp = trans_tbl[trans_tbl['status'] == 'End']

  #Merge Start and End
  df = pd.merge(
      start_grp,
      end_grp,
      on=['project_id', 'operation_id'],
      suffixes=('_start', '_end')
  )
  df['diff'] = df['timestamp_end'] - df['timestamp_start']

  df = df.groupby(['project_id','operation_id']).agg(avg_timestamp=("diff","mean"))
  return df
```

```python
fetch_avg_timestamp_per_grp(records)
```
