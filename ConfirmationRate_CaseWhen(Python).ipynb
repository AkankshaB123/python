```python
def confirmation_rate(signups: pd.DataFrame, confirmations: pd.DataFrame) -> pd.DataFrame:
  df = pd.merge(left = signups['user_id'], right = confirmations[['user_id','action']], how='left', on='user_id')
  df['flag'] = 0
  df.loc[df['action'] == 'confirmed', 'flag'] = 1 #Logic for new column
  df['flag1'] = df['flag']
  #View the correction
  df = df.groupby('user_id').agg({'flag':'count','flag1':'sum'}).reset_index()
  df['confirmation_rate'] = (df['flag1']/df['flag']).round(2).apply(lambda x: '{:.2f}'.format(x))
  return df[['user_id','confirmation_rate']]
```

```python
confirmation_rate(signups, confirmations)
```
