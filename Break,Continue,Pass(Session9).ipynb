# Break, Continue and Pass (For Loop Advanced Topics)
1. Pass - That can help you fill gaps of no execution
2. Continue - does nothing, it goes to back to the next iteration of the loop
3. Break - Breaks out the innermost loop when it gets executed

-----------------------------
Use-cases:
* Used in algorithmic trading - make buy/sell calls - continue
* Also save dat for future for each stock - pass
* if old_pattern(similar to 2008) == TRUE - break (stopping_condition)

# RULE: continue skips; pass ignores; break exits.

```python
for x in range(3):
    if x == 1:
        pass
    print(x)
```

```python
for x in range(3):
    if x == 1:
        continue
    print(x)
```

```python
stocks_list = ['A','B','C','D']
holdings = ['B','C']

for stock in stocks_list:
    if stock not in holdings:
        print(f'Buy {stock}')
    else:
        continue
```

```python
stocks_list = ['A','B','C','D']
holdings = ['B','C']

for stock in stocks_list:
    if stock in holdings:
        continue
    else:
        print(f'{stock}')
        # Send data request to server
        # Process server output in algo function
        # Make decision buy/sell
        print("Declined")
```

```python
stocks_list = ['A','B','C','D']
holdings = ['B','C']

for stock in stocks_list:
    if stock in holdings:
        pass
    else:
        print(f'{stock}')
        # Send data request to server
        # Process server output in algo function
        # Make decision buy/sell
        print("Declined")
```

```python
stocks_list = ['A','B','C','D']
holdings = ['B','C']

for stock in stocks_list:
    if stock in holdings:
        break
    else:
        print(f'{stock}')
        # Send data request to server
        # Process server output in algo function
        # Make decision buy/sell
        print("Declined")
```


* Pass keyword - does nothing, and moves forward in the code
* Continue keyword - also does nothing, but doesn't move forward, it goes to the start of the inner most loop

break, continue & pass

* break - Break Stops the Loop at any point of time when it's executed
* continue - When continue is executed, Python, stops executing anything below
* and takes us back to the statrting of the loop [next iteration]
* pass - It does the same thing Like continue, basically it does nothing
* but it simply passes that case and runs everything below it


# Task
Given a List - [5,15,25,678,333,1005,876,421]
* The Moment you find a number that is divisible by 3 and 5 - Print it - later you should also print the quotient
* Otherwise you go to the next element
* You end the loop when you find a number greater than 1000

```python
l1 = [5,15,25,678,333,1005,876,421]
for i in range(len(l1)):
  print(l1[i])
  if l1[i] > 1000:
    break
  elif (l1[i]%3 != 0) or (l1[i]%5 != 0):
    continue
  else:
    print(l1[i])
    pass
  print("The quotients are {}, {}".format(l1[i]//3, l1[i]//5))
```
