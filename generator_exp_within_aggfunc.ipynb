```python
# Taking input
n = int(input())

def difference_sum_squares(n):
# Write your code here
    sum_of_squares = sum(i**2 for i in range(n+1))
    square_of_sum = sum(range(1,n+1))**2
    difference_sum_squares = square_of_sum-sum_of_squares
    return difference_sum_squares
# Print the output
print(difference_sum_squares(n))
```
