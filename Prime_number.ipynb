```python
# Taking input
num = int(input())

def is_prime(n):
    # Write your code
    if n < 2:
        return "Not Prime"
    i=2
    output = "Prime"
    res=1
    while i*i<=n:
        if(n%i==0):
            res=i
            output = "Not Prime"
        else:
            res=i
            output = "Prime"
        i=i+1
    return output

# Print the output
print(is_prime(num))
```

* 1 is not a Prime Number
* Prime no is only divisible by itself
* Because of i*i<=n helps going into the else block: Prime number, because otherwise it's just 7 being a prime number, 7/7 output as not prime
* Input = 9, checks for 3 because i=3, 3*3<=9, last output="Not Prime"
