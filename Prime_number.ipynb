```python
# Taking input
num = int(input())

def is_prime(n):
    # Write your code
    i=2
    output = "Prime"
    res=1
    while i*i<=n:
        if(n%i==0):
            res=i
            output = "Not Prime"
        else:
            res=i
            output = "Prime"
        i=i+1
    return output

# Print the output
print(is_prime(num))
```
