```python
a = [[1,2,3,4]]
b = [[5,6,7,8]]
```

```python
print(a)
```

```python
a.extend(b)
```

```python
print(a)
```

Here you're modifying the same list twice, so the second result includes the effect of the first.

```python
print('Input a:',a)
print('Input b:',b)
x = a.copy()
x.append(b)
print('append:',x)
y = a.copy()
y.extend(b)
print('extend:',y)
```

```python
a1 = [1,2,3,4]
b1 = [5,6,7,8]
print('Input a:',a1)
print('Input b:',b1)
x = a1.copy()
x.append(b1)
print('append:',x)
y = a1.copy()
y.extend(b1)
print('extend:',y)
```
