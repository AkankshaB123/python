```python
a = [[1,2,3,4]]
b = [[5,6,7,8]]
```

```python
print(a)
```

```python
a.extend(b)
```

```python
print(a)
```

Here you're modifying the same list twice, so the second result includes the effect of the first.

```python
print('Input a:',a)
print('Input b:',b)
x = a.copy()
x.append(b)
print('append:',x)
y = a.copy()
y.extend(b)
print('extend:',y)
```

```python
print('x:',x)
print('x[0]',x[0])
print('x[1]',x[1])
print('x[0][0]',x[0][0])
print('x[1][0]',x[1][0])
print('x[1][0][1]',x[1][0][1]) #3 list in append
```

# 3 list in append and 2 list in append

Notice the difference
1. append - adds both the item in a new list [[]]
2. extend - adds items to a

```python
print('y:',y)
print('y[0]',y[0])
print('y[1]',y[1])
print('y[0][0]',y[0][0])
print('y[1][1]',y[1][1]) #2 list in extend
```

```python
a1 = [1,2,3,4]
b1 = [5,6,7,8]
print('Input a:',a1)
print('Input b:',b1)
x = a1.copy()
x.append(b1)
print('append:',x)
y = a1.copy()
y.extend(b1)
print('extend:',y)
```
