# Conditional Statements in Python

* condition - equal to, greater than, less than, in, not in
* connector - and, or
* action associated with the above
--------------------------------------------------------------------
* if - 1 time (necessary), else - 1 time
* elif - multiple times
--------------------------------------------------------------------

#Question:
1. Male, Female
2. 20, 40, 40+
3. Gender = ['m', 'M', 'Male','f','F','Female']
4. age = number

```python
gender = input("Enter your gender: ")
age = int(input("Enter your age: "))
if gender in('m','M','Male','male'):
    gender = "Male"
    if ((age <= 20)):
      age = "20"
    elif ((age > 20) and (age <= 40)):
      age = "20-40 age group"
    elif (age>40):
      age = "41+"
    print(f"The person is {gender} with {age} of age")
elif gender in ('f','F','Female','female'):
    gender = "Female"
    if ((age <= 20)):
      age = "20"
    elif ((age > 20) and (age <= 40)):
      age = "20-40 age group"
    elif (age>40):
      age = "41+"
    print(f"The person is {gender} with {age} of age")
else:
   print("Please enter proper format for gender")
```
